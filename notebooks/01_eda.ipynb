```python
import glob

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
files = glob.glob('../data/raw/*.csv')

dataframes = []

for file in files:
    df = pd.read_csv(file)
    df['source_file'] = file.split('/')[-1]
    dataframes.append(df) 

apartments = pd.concat(dataframes, ignore_index=True)
apartments.head()
```

```python
print('shape: ',apartments.shape)
print('\n')
print('data types:\n',apartments.dtypes)
```

```python
print('missing percent:\n', (apartments.isnull().sum()/len(apartments) * 100).sort_values(ascending=False))
```

```python
cols_to_check = [col for col in apartments.columns if col != 'source_file']
print('duplicated values: ', apartments.duplicated(subset=cols_to_check).sum())
print('duplicated values percent: ', round(apartments.duplicated(subset=cols_to_check).sum()/len(apartments) * 100))
```
